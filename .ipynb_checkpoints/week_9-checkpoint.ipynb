# Generators and iterators

```python
# creating a basic iterator from an iterable
sports = [ "baseball", "soccer", "footbal", "hockey", "baseball" ]
my_iter = iter(sports)
print( next(my_iter) )        # outputs first item
print( next(my_iter) )        # outputs second item
for item in my_iter:
    print(item)
print( next(my_iter) )         # will produce an error
```

```python
# creating our own iterator
class Alphabet():
    def __iter__(self):
        self.letters = "abcdefghijklmnopqrstuvwxyz"
        self.index = 0
        return self

    def __next__(self):
        if self.index <= 25:
            char = self.letters[ self.index ]
            self.index += 1
            return char

        else:
            raise StopIteration

for char in Alphabet():
    print(char)
```

# Generators are simplified iterators

```python
# Example
# creating our own range generator with start, stop, and step parameters
def myRange(stop, start=0, step=1):
    while start < stop:
        print( "Generator Start Value: { }".format(start) )
        yield start
        start += step           # increment start, otherwise infinite loop
    for x in myRange(5):
        print( "For Loop X Value: { }".format(x) )
```

```python
# Exercise
#Reverse Iteration: Create an iterator that takes in a list, and when iterated over, it returns the information in a reverse order. Hint: When accepting arguments into an iterator, you need to use the init method, as well as iter and next. The following call should result in “5, 4, 3, 2, 1”.
#for i in RevIter( [ 1, 2, 3, 4, 5 ] ):

# creating an iterator that will iterate an iterable in reverse direction

class ReverseIter():
    def __init__(self, items):
        self.items = items
        self.i = len(items) - 1
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.i >= 0:
            item = self.items[self.i]
            self.i -= 1
            return item
        else:
            raise StopIteration
            
rev_iter = ReverseIter([1, 2, 3, 4, 5])


for num in rev_iter:
    print(num)
```

```python
# Exercise
#Squares: Create a generator that acts like the range function, except it yields a squared number every time. The result of the following call should be “0, 1, 4, 16”.
#for i in range(4):

def squared(n):
    for i in range(n + 1):
        yield i**2
        
for x in squared(4):
    print(x)
```

# Decorators

```python
# creating and applying our own decorator using the @ symbol
def decorator(func):
    def wrap():
        print("===========")
        func()
        print("===========")
    return wrap
@decorator
def printName():
    print("dev_shinobi")
printName()
```

```python
# Decorators with Parameters
# creating a decorator that takes in parameters
def run_times(num):
    def wrap(func):
        for i in range(num):
            func()
    return wrap
@run_times(5)
def sayHello():
    print("Hello")
```

```python
# Functions with Decorators and Parameters
# creating a decorator for a function that accepts parameters
def birthday(func):
    def wrap(name, age):
        func(name, age + 1)
    return wrap
@birthday
def celebrate(name, age):
    print( "Happy birthday {}, you are now {}.".format(name, age) )
celebrate("Paul", 43)
```

```python
# Example
def login_required(func):
    def wrap(user):
        password = input("What is the password?")
        if password == user["password"]:
            func(user)
        else:
            print("Access Denied")
    return wrap
@login_required
def restrictedFunc(user):
    print("Access granted, welcome {}".format(user[ "name" ]) )
user = { "name" : "jess", "password" : "abcdef" }
restrictedFunc(user)
```

```python
# real world sim, restricting function access
def login_required(func):
    def wrap(user):
        password = input("What is the password? ")
        if password == user["password"]:
            func(user)
        else:
            print("Access Denied")
    return wrap
@login_required
def restrictedFunc(user):
    print("Access granted, welcome {}".format(user["name"]))
    
user = {"name" : "Jess", "password" : "ilywpf"}
restrictedFunc(user)
```

```python
# Exercise
def decorator(func):
    def wrap():
        user = int(input("Enter a Number"))
        if user < 100:
            func()
    return wrap
@decorator
def numbers():
    print("Less than 100")

numbers()
```

```python
# Exercise
def route(endpoint):
    def wrap(f):
        print(endpoint)
        f()
    return wrap

@route("/index")
def index():
    print("This is how web pages are made in Flask")
```

# Modules

```python
# import the entire math module
import math
print( math.floor(2.5) )     # rounds down
print( math.ceil(2.5) )      # rounds up
print(math.pi)
```

```python
# Example
# importing only variable and functions rathar than an entire module, better efficiency
from math import floor, pi
print( floor(2.5) )
# print( ceil(2.5) )   will cause an error because we only imported floor and pi, not ceil and not all of math
print(pi)
```

```python
# using an Alias
# using the 'as' keyword to create an alias for imports
from math import floor as f
print( f(2.5) )
```
